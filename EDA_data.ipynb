## Data quality and interpretation

Three numerical anomalies are flagged below: ORD0052 has a negative quantity, ORD0071 has a discount above 100%, and ORD0091 has a recorded amount of 99,999. The existing analysis excludes only ORD0091 from order-value comparisons. This is an exploratory choice, not a complete validation policy; the other anomalies remain and may affect results.

Category totals use recorded labels. Product-to-category consistency needs further validation. No source-backed corrections are applied.

```python
df.describe(include=['object','string'])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
rev = df[df['Total_Amount'] < 99999]

rev.groupby('Category')['Total_Amount'].sum().sort_values(ascending=False).plot(kind='bar', title='Recorded Order Value by Category (excludes ORD0091)')
plt.show()

rev.groupby('City')['Total_Amount'].sum().sort_values(ascending=False).plot(kind='bar', title='Recorded Order Value by City (excludes ORD0091)')
plt.show()

pd.crosstab(df['Payment_Method'], df['Order_Status'], normalize='index').plot(kind='bar', stacked=True, title='Order Status by Payment Method')
plt.show()
```

```python
rev = df[df['Total_Amount'] < 99999]
print(rev.groupby('Category')['Total_Amount'].sum().sort_values(ascending=False).round(0))
print(rev.groupby('City')['Total_Amount'].sum().sort_values(ascending=False).round(0))
print(pd.crosstab(df['Payment_Method'], df['Order_Status']))
```

```python
delivered = df[df['Order_Status'] == 'Delivered']
delivered.groupby('Category')['Total_Amount'].sum().sort_values(ascending=False).plot(kind='bar', title='Recorded Value of Delivered Orders by Category')
plt.show()
```

```python
monthly = rev.set_index('Order_Date').resample('ME')['Total_Amount'].sum()
monthly.plot(marker='o', title='Monthly Recorded Order Value (excludes ORD0091)')
plt.show()
```

## Limitations and next steps

The sample contains 116 orders. Group differences describe this sample and do not establish causation. The latest recorded date is 25 September 2026; September coverage is incomplete or unverified, so its monthly total is not directly comparable with full months. The dataset alone cannot explain the change.

Before making business decisions, verify anomaly values, product categories, dataset provenance and reporting coverage. Delivered status is not proof of collected payment. Read `README.md` for findings and reproduction instructions.
