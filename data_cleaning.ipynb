```python
for col in ['City','Category','Payment_Method','Order_Status','Product_Name']:
    print(col,df[col].unique(),'\n')
```

```python
for c in ["Customer_Name", "City", "Product_Name", "Category", "Order_Status", "Payment_Method"]:
    df[c] = df[c].str.strip().str.title()
df["Payment_Method"] = df["Payment_Method"].replace({"Upi": "UPI"})
df
```

```python
df = df.drop_duplicates().dropna(subset=["Order_ID", "Customer_ID"]).reset_index(drop=True)
df
```

```python
df["Customer_Name"] = df["Customer_Name"].fillna(df.groupby("Customer_ID")["Customer_Name"].transform("first"))
df["Unit_Price"] = df["Unit_Price"].fillna(df.groupby("Product_ID")["Unit_Price"].transform("median"))
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].mode()[0]).astype(int)
```

```python
df.info()
print("Missing:", df.isna().sum().sum(), "| Duplicates:", df.duplicated().sum())
```

```python
df[df['Order_Date'].isna()]
```

```python
df = df.dropna(subset=['Order_Date']).reset_index(drop=True)
```

```python
df["Calc_Total"] = (df["Quantity"] * df["Unit_Price"] * (1 - df["Discount"] / 100)).round(2)
print("Total_Amount mismatches:", (df["Calc_Total"] != df["Total_Amount"].round(2)).sum())
```

```python
df = df.drop(columns="Calc_Total")

print("Shape:", df.shape)
print("Missing:", df.isna().sum().sum(), "| Duplicates:", df.duplicated().sum())

df.to_csv("cleaned_data.csv", index=False)
```
